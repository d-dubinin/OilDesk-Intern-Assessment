# Python Basics and Data Manipulation

This notebook loads the commodity price CSV, parses and cleans it, filters for selected commodities and the year 2021, computes monthly averages, and visualises the results.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

## 1. Load the CSV & Inspection

The raw file has 7 header rows of metadata before the actual data begins. We skip those rows and assign clean column names manually.

The date column uses `DD/MM/YYYY` format, so we pass `dayfirst=True` to `pd.to_datetime`.

```python
# Skip the 7 metadata rows; row 8 in the file is the first data row
COLUMNS = ["date", "copper", "aluminium", "zinc", "lead", "tin", "crude_oil"]

df = pd.read_csv(
    "../data/MarketData.csv",
    skiprows=7,
    header=None,
    names=COLUMNS,
)

# Parse dates
df["date"] = pd.to_datetime(df["date"], dayfirst=True)

# Coerce all price columns to numeric
price_cols = ["copper", "aluminium", "zinc", "lead", "tin", "crude_oil"]
df[price_cols] = df[price_cols].apply(pd.to_numeric, errors="coerce")

print(
    f"Loaded {len(df):,} rows | Date range: {df['date'].min().date()} → {df['date'].max().date()}"
)
df.head()
```

```python
df.info()
```

```python
df[price_cols].describe().round(2)
```

**Dataset observations:**

- 3,391 rows covering January 2010 to December 2022 — no missing values in any column across the full period.
- Crude oil shows a minimum of **−$37.63**, reflecting the historic WTI negative settlement on 20 April 2020 when storage capacity hit physical limits during COVID lockdowns. This is the only negative price in the dataset and is kept as-is — it is a real market observation, not a data error.
- Across the 13-year dataset, copper averaged $7,073/t. As we will see, 2021 prices were significantly above this long-run mean.

## 2. Filter for Selected Commodities and Year 2021

We assess Copper, Zinc, and Crude Oil.

Since each commodity is its own column (wide format), we:
1. Keep only the relevant columns
2. Filter rows where the year is 2021
3. Melt to long format so each row is one (date, commodity, price) observation as this makes grouping and plotting much cleaner

```python
SELECTED = ["copper", "zinc", "crude_oil"]

# Filter year
df_2021 = df[df["date"].dt.year == 2021][["date"] + SELECTED].copy()

# Melt to long format
df_long = df_2021.melt(
    id_vars="date", value_vars=SELECTED, var_name="commodity", value_name="price"
)
df_long.head(10)
```

```python
df_2021[SELECTED].describe().round(2)
```

### Daily Settlement Prices — 2021

Before aggregating to monthly averages it is useful to see the raw daily series to understand the shape of each commodity's move through the year.

```python
LABELS = {
    "copper": "Copper (USD/t)",
    "zinc": "Zinc (USD/t)",
    "crude_oil": "Crude Oil (USD/bbl)",
}

fig, axes = plt.subplots(3, 1, figsize=(12, 8), sharex=True)

for ax, commodity in zip(axes, SELECTED):
    data = df_long[df_long["commodity"] == commodity].sort_values("date")
    ax.plot(data["date"], data["price"], linewidth=0.9)
    ax.set_title(LABELS[commodity])
    ax.spines[["top", "right"]].set_visible(False)
    ax.grid(axis="y", linestyle="--", alpha=0.3)

fig.suptitle("Daily Settlement Price — 2021")
fig.tight_layout()
plt.show()
```

## 3. Calculate Monthly Average Price

We group by commodity and month, then take the mean settlement price.

`pd.Grouper(key='date', freq='ME')` groups by month-end period cleanly without needing to manually extract year/month.

```python
monthly_avg = (
    df_long.groupby(["commodity", pd.Grouper(key="date", freq="ME")])["price"]
    .mean()
    .reset_index()
)

monthly_avg["month"] = monthly_avg["date"].dt.strftime("%b")
monthly_avg["price"] = monthly_avg["price"].round(2)

monthly_avg.head(12)
```

## 4. Table View

Pivot to wide format for a clean side-by-side comparison.

```python
table = monthly_avg.pivot(index="month", columns="commodity", values="price")

# Reorder months in chronological order
month_order = [
    "Jan",
    "Feb",
    "Mar",
    "Apr",
    "May",
    "Jun",
    "Jul",
    "Aug",
    "Sep",
    "Oct",
    "Nov",
    "Dec",
]
table = table.reindex(month_order)

table
```

## 5. Chart — Monthly Average Price per Commodity (2021)

Three subplots, one per commodity. Crude Oil uses a separate y-axis scale (USD/barrel) vs the metals (USD/tonne), so separating them avoids a misleading chart.

```python
fig, axes = plt.subplots(3, 1, figsize=(10, 7), sharex=True)

LABELS = {
    "copper": "Copper (USD/tonne)",
    "zinc": "Zinc (USD/tonne)",
    "crude_oil": "Crude Oil (USD/bbl)",
}

for ax, commodity in zip(axes, ["copper", "zinc", "crude_oil"]):
    data = monthly_avg[monthly_avg["commodity"] == commodity].sort_values("date")
    ax.plot(data["date"], data["price"], marker="o")
    ax.set_title(LABELS[commodity])

fig.suptitle("Monthly Average Price — 2021")
fig.tight_layout()
plt.show()
```

## 6. Observations

**Copper** had a strong 2021 averaging $9,292/t above its 13-year mean of $7,073/t. Prices climbed from $7,967/t in January to a peak of $10,162/t in May, driven by post-COVID demand recovery and supply constraints, then consolidated in the $9,300–$9,650 range for the remainder of the year.

**Zinc** moved more gradually, rising from $2,730/t in January to $3,371/t in December; a gain of 23% through the year with the sharpest move coming in Q4.

**Crude Oil** recovered throughout the year, going from $52/bbl in January to $81/bbl in October before easing slightly toward year-end. This represents a full recovery from the events of 2020, when WTI settled at negative price of −$37.63/bbl.

Overall, all three commodities moved in the same direction in 2021.
